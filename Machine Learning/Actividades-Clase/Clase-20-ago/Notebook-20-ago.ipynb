**Data imputation** es una técnica muy común que consiste en el proceso de reemplazar valores faltantes o nulos en un conjunto de datos con sustitutos estimados o plausibles (media, moda, mediana, etc.). Es importante saber cuando utilizarla porque puede generar **data leakage** porque al hacer uso de información artificial durante el entrenamiento del modelo aumenta la posibilidad de que las predicciones que haga no sean útiles, ya que se hace dependiente a información que no estará disponible al momento de hacer predicciones reales.

## 4. Cambiar las variables categoricas a representaciones númericas

Los números asignados no representan una jerarquía, si no que se usan para diferencias los diferentes valores de cada variable para poder ser analizados más fácilmente por el modelo.

```python
from sklearn.preprocessing import LabelEncoder
from IPython.display import display
import pandas as pd

df_encoded = df1.copy()

categorical_columns = ["VehicleManufacturerName", "TestedTransmissionType", "DriveSystemDescription", "TestCategory", "TestFuelTypeDescription", "AftertreatmentDevice"]

label_encoders = {}
mapping_tables = {}

for col in categorical_columns:
    le = LabelEncoder()
    # Replace original with encoded values
    df_encoded[col] = le.fit_transform(df_encoded[col].astype(str))
    # Save encoder
    label_encoders[col] = pd.DataFrame({
        "Category": le.classes_,
        "Encoded Value": le.transform(le.classes_)
    })

# Display mappigs
for col, mapping in mapping_tables.items():
    print(f"\nMapping for {col}:")
    display(mapping)

df_encoded
```
